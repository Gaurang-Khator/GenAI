```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model=init_chat_model("groq:qwen/qwen3-32b")
model
```

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    title:str = Field(description="The title of the movie")
    rating:float = Field(description="The rating of the movie out of 10")
    director:str = Field(description="The director of the movie")
    year:int = Field(description="The release year of the movie")
    genre:list[str] = Field(description="The genre of the movie")
    cast:list[str] = Field(description="The main cast of the movie")
```

```python
model.invoke("Provide details about the movie London Has Fallen")
```

```python
model_with_structure = model.with_structured_output(Movie)
model_with_structure.invoke("Provide details about the movie London Has Fallen")
```

```python
class Actor(BaseModel):
    name:str
    role:str

class MovieDetails(BaseModel):
    title:str
    rating:float 
    director:str 
    genre:list[str]
    cast:list[Actor]

model_with_structure=model.with_structured_output(MovieDetails)
model_with_structure.invoke("Provide details about the movie Mission Impossible")
```
