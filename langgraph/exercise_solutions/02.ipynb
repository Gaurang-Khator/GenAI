```python
from typing import TypedDict, List
from langgraph.graph import StateGraph 
import math
```

```python
class AgentState(TypedDict):
    name : str
    values : List[int]
    operation : str
    result: str
```

```python
def compute_operation(state: AgentState) -> AgentState:
    """Simple node that returns the answer based on operation."""

    if(state['operation'] == "+"):

        state['result'] = f"Hi {state['name']}, your answer is: {sum(state['values'])}"

    elif(state['operation'] == "*"):

        state['result'] = f"Hi {state['name']}, your answer is: {math.prod(state['values'])}"

    
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("compute", compute_operation)

graph.set_entry_point("compute")
graph.set_finish_point("compute")

app = graph.compile()
```

```python
ans = app.invoke({"name": "Gaurang", "values": [1,2,3,4], "operation": "*"})
```

```python
print(ans['result'])
```
