```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    values : List[int]
    name : str
    result : str
```

```python
def process_values(state: AgentState) -> AgentState:
    """Simple node that processes the values."""

    print(state)

    state['result'] = f"Hey {state['name']}, your total is {sum(state['values'])}"

    print(state)
    
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("processor", process_values)

graph.set_entry_point("processor")
graph.set_finish_point("processor")

app = graph.compile()
```

```python
ans = app.invoke({"values" : [1, 2, 3, 4, 5], "name" : "Gaurang"})
```

```python
print(ans)
print("====")
print(ans['result'])
```
